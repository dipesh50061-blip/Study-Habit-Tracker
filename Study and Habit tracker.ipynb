```python
import json
from datetime import datetime

name = input("Enter Your Name: ")
print("Welcome to The Study and Habit Tracker,", name)

session = []
habit = []

# =========================
# STUDY FUNCTIONS
# =========================

def add_study_sessions():

    print("\n--- Add Study Sessions ---")

    sub_name = input("Enter Subject name: ")

    time_value = int(input("Enter study time: "))
    time_unit = input("Enter unit (h for hours / m for minutes): ").lower()

    if time_unit == "h":
        timing = time_value * 60

    elif time_unit == "m":
        timing = time_value

    else:
        print("Invalid Unit")
        return

    current_time = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    sessions = {
        "Subject": sub_name,
        "Timing": timing,
        "Timestamp": current_time
    }

    session.append(sessions)

    save_sessions()

    print("Subject Added Successfully")


def view_study_sessions():

    print("\n--- View Study Sessions ---")

    if len(session) == 0:
        print("No Session Found")

    else:
        for index, sessions in enumerate(session):

            hours = sessions["Timing"] // 60
            minutes = sessions["Timing"] % 60

            print("\nSession", index + 1)
            print("Subject:", sessions["Subject"])
            print("Timing:", hours, "hr", minutes, "min")
            print("Created:", sessions["Timestamp"])


def total_study_time():

    print("\n--- Total Study Time ---")

    total = 0

    for sessions in session:
        total += sessions["Timing"]

    hours = total // 60
    minutes = total % 60

    print("Total Study Time:", hours, "hr", minutes, "min")


def delete_study_sessions():

    print("\n--- Delete Study Sessions ---")

    if len(session) == 0:
        print("No Sessions to Delete")

    else:

        for index, sessions in enumerate(session):
            print(index + 1, sessions["Subject"])

        delete_num = int(input("Enter Session Number to delete: "))

        if delete_num >= 1 and delete_num <= len(session):

            deleted = session.pop(delete_num - 1)

            save_sessions()

            print(deleted["Subject"], "deleted successfully!")

        else:
            print("Invalid Session Number")


def edit_study_sessions():

    print("\n--- Edit Study Sessions ---")

    if len(session) == 0:
        print("No Sessions to Edit")

    else:

        for index, sessions in enumerate(session):
            print(index + 1, sessions["Subject"])

        edit_num = int(input("Enter Session number to Edit: "))

        if edit_num >= 1 and edit_num <= len(session):

            new_subject = input("Enter new Subject name: ")

            time_value = int(input("Enter new study time: "))
            time_unit = input("Enter unit (h for hours / m for minutes): ").lower()

            if time_unit == "h":
                new_timing = time_value * 60

            elif time_unit == "m":
                new_timing = time_value

            else:
                print("Invalid Unit")
                return

            session[edit_num - 1]["Subject"] = new_subject
            session[edit_num - 1]["Timing"] = new_timing

            save_sessions()

            print("Session updated successfully")

        else:
            print("Invalid session number")


def search_study_sessions():

    print("\n--- Search Study Sessions ---")

    keyword = input("Enter subject to search: ").lower()

    found = False

    for sessions in session:

        if keyword in sessions["Subject"].lower():

            hours = sessions["Timing"] // 60
            minutes = sessions["Timing"] % 60

            print("\nSubject:", sessions["Subject"])
            print("Timing:", hours, "hr", minutes, "min")
            print("Created:", sessions["Timestamp"])

            found = True

    if found == False:
        print("No matching session found")


def sort_study_sessions():

    print("\n--- Sort Study Sessions ---")

    if len(session) == 0:
        print("No Sessions Found")

    else:

        sorted_sessions = sorted(
            session,
            key=lambda x: x["Timing"],
            reverse=True
        )

        for index, sessions in enumerate(sorted_sessions):

            hours = sessions["Timing"] // 60
            minutes = sessions["Timing"] % 60

            print("\nSession", index + 1)
            print("Subject:", sessions["Subject"])
            print("Timing:", hours, "hr", minutes, "min")
            print("Created:", sessions["Timestamp"])


# =========================
# HABIT FUNCTIONS
# =========================

def add_habits():

    print("\n--- Add Habits ---")

    habit_name = input("Enter Your Habit: ")

    time_value = int(input("Enter habit time: "))
    time_unit = input("Enter unit (h for hours / m for minutes): ").lower()

    if time_unit == "h":
        timing = time_value * 60

    elif time_unit == "m":
        timing = time_value

    else:
        print("Invalid Unit")
        return

    current_time = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    habits = {
        "Habit": habit_name,
        "Time Per Habit": timing,
        "Timestamp": current_time
    }

    habit.append(habits)

    save_habits()

    print("Habit Added Successfully")


def view_habits():

    print("\n--- View Habits ---")

    if len(habit) == 0:
        print("No Habit Found")

    else:
        for index, habits in enumerate(habit):

            hours = habits["Time Per Habit"] // 60
            minutes = habits["Time Per Habit"] % 60

            print("\nHabit", index + 1)
            print("Habit:", habits["Habit"])
            print("Timing:", hours, "hr", minutes, "min")
            print("Created:", habits["Timestamp"])


def total_habit_time():

    print("\n--- Total Habit Time ---")

    total = 0

    for habits in habit:
        total += habits["Time Per Habit"]

    hours = total // 60
    minutes = total % 60

    print("Total Habit Time:", hours, "hr", minutes, "min")


def delete_habit_sessions():

    print("\n--- Delete Habit Sessions ---")

    if len(habit) == 0:
        print("No Habit to Delete")

    else:

        for index, habits in enumerate(habit):
            print(index + 1, habits["Habit"])

        delete_number = int(input("Enter Habit Number to delete: "))

        if delete_number >= 1 and delete_number <= len(habit):

            deletes = habit.pop(delete_number - 1)

            save_habits()

            print(deletes["Habit"], "deleted successfully!")

        else:
            print("Invalid Habit Number")


def edit_habit_sessions():

    print("\n--- Edit Habit ---")

    if len(habit) == 0:
        print("No Habit to Edit")

    else:

        for index, habits in enumerate(habit):
            print(index + 1, habits["Habit"])

        edit_number = int(input("Enter Habit Number to Edit: "))

        if edit_number >= 1 and edit_number <= len(habit):

            new_habit = input("Enter new Habit name: ")

            time_value = int(input("Enter new habit time: "))
            time_unit = input("Enter unit (h for hours / m for minutes): ").lower()

            if time_unit == "h":
                new_timing = time_value * 60

            elif time_unit == "m":
                new_timing = time_value

            else:
                print("Invalid Unit")
                return

            habit[edit_number - 1]["Habit"] = new_habit
            habit[edit_number - 1]["Time Per Habit"] = new_timing

            save_habits()

            print("Habit updated successfully")

        else:
            print("Invalid habit number")


def search_habits():

    print("\n--- Search Habits ---")

    keyword = input("Enter habit to search: ").lower()

    found = False

    for habits in habit:

        if keyword in habits["Habit"].lower():

            hours = habits["Time Per Habit"] // 60
            minutes = habits["Time Per Habit"] % 60

            print("\nHabit:", habits["Habit"])
            print("Timing:", hours, "hr", minutes, "min")
            print("Created:", habits["Timestamp"])

            found = True

    if found == False:
        print("No matching habit found")


def sort_habits():

    print("\n--- Sort Habits ---")

    if len(habit) == 0:
        print("No Habits Found")

    else:

        sorted_habits = sorted(
            habit,
            key=lambda x: x["Time Per Habit"],
            reverse=True
        )

        for index, habits in enumerate(sorted_habits):

            hours = habits["Time Per Habit"] // 60
            minutes = habits["Time Per Habit"] % 60

            print("\nHabit", index + 1)
            print("Habit:", habits["Habit"])
            print("Timing:", hours, "hr", minutes, "min")
            print("Created:", habits["Timestamp"])


# =========================
# SAVE / LOAD FUNCTIONS
# =========================

def save_sessions():

    with open("session.json", "w") as file:
        json.dump(session, file)


def save_habits():

    with open("habit.json", "w") as file:
        json.dump(habit, file)


def load_sessions():

    global session

    try:

        with open("session.json", "r") as file:
            session = json.load(file)

    except FileNotFoundError:
        session = []


def load_habits():

    global habit

    try:

        with open("habit.json", "r") as file:
            habit = json.load(file)

    except FileNotFoundError:
        habit = []


# AUTO LOAD
load_sessions()
load_habits()

# =========================
# MAIN MENU
# =========================

while True:

    print("\n===== Study and Habit Tracker =====")
    print("1. Add Study Sessions")
    print("2. View Study Sessions")
    print("3. Add Habits")
    print("4. View Habits")
    print("5. Total Study Time")
    print("6. Total Habit Time")
    print("7. Delete Study Sessions")
    print("8. Delete Habit Sessions")
    print("9. Edit Study Sessions")
    print("10. Edit Habit")
    print("11. Search Study Sessions")
    print("12. Search Habits")
    print("13. Sort Study Sessions")
    print("14. Sort Habits")
    print("15. Exit")

    choice = input("Enter Number: ")

    if choice == "1":
        add_study_sessions()

    elif choice == "2":
        view_study_sessions()

    elif choice == "3":
        add_habits()

    elif choice == "4":
        view_habits()

    elif choice == "5":
        total_study_time()

    elif choice == "6":
        total_habit_time()

    elif choice == "7":
        delete_study_sessions()

    elif choice == "8":
        delete_habit_sessions()

    elif choice == "9":
        edit_study_sessions()

    elif choice == "10":
        edit_habit_sessions()

    elif choice == "11":
        search_study_sessions()

    elif choice == "12":
        search_habits()

    elif choice == "13":
        sort_study_sessions()

    elif choice == "14":
        sort_habits()

    elif choice == "15":

        save_sessions()
        save_habits()

        print("All data auto-saved successfully!")
        print("Goodbye!")

        break

    else:
        print("Invalid Input")
```
